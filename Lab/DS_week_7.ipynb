```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris

# Load Iris dataset
iris = load_iris()

df = pd.DataFrame(iris.data, columns=iris.feature_names)

# Add species column
df['species'] = iris.target

print("First 5 rows:")
print(df.head())
```

```python
# Histogram

plt.figure(figsize=(6,4))

sns.histplot(df['sepal length (cm)'], bins=20, kde=True)

plt.title("Histogram of Sepal Length")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Frequency")

plt.show()
```

```python
# Boxplot

plt.figure(figsize=(6,4))

sns.boxplot(x=df['sepal length (cm)'])

plt.title("Boxplot of Sepal Length")
plt.xlabel("Sepal Length (cm)")

plt.show()
```

```python
# Pie Chart

species_counts = df['species'].value_counts()

plt.figure(figsize=(6,6))

plt.pie(
    species_counts,
    labels=iris.target_names,
    autopct='%1.1f%%'
)

plt.title("Species Distribution")

plt.show()
```

```python
# Scatterplot

plt.figure(figsize=(6,4))

sns.scatterplot(
    x=df['sepal length (cm)'],
    y=df['sepal width (cm)'],
    hue=df['species']
)

plt.title("Sepal Length vs Sepal Width")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Sepal Width (cm)")

plt.show()
```

```python
# Line Chart

plt.figure(figsize=(6,4))

plt.plot(df['sepal length (cm)'])

plt.title("Line Chart of Sepal Length")
plt.xlabel("Sample Index")
plt.ylabel("Sepal Length (cm)")

plt.show()
```

```python
# Bar Chart

plt.figure(figsize=(6,4))

sns.barplot(
    x=df['species'],
    y=df['sepal length (cm)']
)

plt.title("Mean Sepal Length by Species")

plt.xticks(
    [0, 1, 2],
    iris.target_names
)

plt.xlabel("Species")
plt.ylabel("Mean Sepal Length (cm)")

plt.show()
```

```python
# Heatmap

plt.figure(figsize=(8,6))

sns.heatmap(
    df.iloc[:, :4].corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Feature Correlation Heatmap")

plt.show()
```

```python
# Bubble Chart

plt.figure(figsize=(6,4))

plt.scatter(
    df['sepal length (cm)'],
    df['sepal width (cm)'],
    s=df['petal length (cm)'] * 20,
    alpha=0.5,
    c=df['species']
)

plt.title("Bubble Chart")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Sepal Width (cm)")

plt.show()
```

```python
# Pair Plot

sns.pairplot(df.iloc[:, :4])

plt.suptitle("Pair Plot of Iris Features", y=1.02)

plt.show()
```

```python
# Violin Plot

plt.figure(figsize=(6,4))

sns.violinplot(
    x=df['species'],
    y=df['sepal length (cm)']
)

plt.title("Violin Plot: Sepal Length by Species")

plt.xticks(
    [0, 1, 2],
    iris.target_names
)

plt.xlabel("Species")
plt.ylabel("Sepal Length (cm)")

plt.show()
```
