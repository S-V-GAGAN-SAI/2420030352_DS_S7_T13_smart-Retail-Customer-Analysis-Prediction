```python
# Python String Methods 

a = " Hello World "

print(a)                              # Original string

print(a.upper())                      # Convert to uppercase
print(a.lower())                      # Convert to lowercase
print(a.title())                      # First letter of each word uppercase
print(a.capitalize())                 # First letter uppercase
print(a.swapcase())                   # Swap uppercase and lowercase

print(a.strip())                      # Remove spaces from both sides
print(a.lstrip())                     # Remove spaces from left
print(a.rstrip())                     # Remove spaces from right

print(a.replace("World", "Python"))   # Replace substring
print(a.split())                      # Split string into list
print("-".join(["A", "B", "C"]))      # Join list into string

print(a.find("World"))                # Find index of substring (-1 if not found)
print(a.index("World"))               # Find index (error if not found)
print(a.count("o"))                   # Count occurrences of character

print(a.startswith(" "))              # Check starting characters
print(a.endswith(" "))                # Check ending characters

print("Hello".isalpha())              # True if only alphabets
print("12345".isdigit())              # True if only digits
print("abc123".isalnum())             # True if letters and digits
print("   ".isspace())                # True if only spaces
print("hello".islower())              # True if all lowercase
print("HELLO".isupper())              # True if all uppercase
print("Hello World".istitle())        # True if title case

print("Hi".center(10, "*"))           # Center align with '*'
print("Hi".ljust(10, "-"))            # Left justify
print("Hi".rjust(10, "-"))            # Right justify
print("25".zfill(5))                  # Pad with leading zeros

print(len(a))                         # Length of string
print(a[::-1])                        # Reverse string
print(a[0:5])                         # Slicing
print(a[-1])                          # Last character
print(a[-5:])                         # Last 5 characters
```
