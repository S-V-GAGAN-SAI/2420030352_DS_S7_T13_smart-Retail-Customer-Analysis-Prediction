```python
import matplotlib.pyplot as plt

y1 = []
y2 = []

x = range(-100, 100, 10)

for i in x:
    y1.append(i**2)

for i in x:
    y2.append(-i**2)

plt.plot(x, y1)
plt.plot(x, y2)

plt.xlabel("x")
plt.ylabel("y")
plt.ylim(-2000, 2000)

plt.axhline(0)
plt.axvline(0)

plt.savefig("quad.png")
plt.show()
```

```python
from sklearn.datasets import load_iris
import pandas as pd

iris = load_iris()

df = pd.DataFrame(
    iris.data,
    columns=[
        'sepal length (cm)',
        'sepal width (cm)',
        'petal length (cm)',
        'petal width (cm)'
    ]
)

print(df.head())
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

iris = sns.load_dataset("iris")

setosa = iris.loc[iris.species == "setosa"]

sns.kdeplot(
    x=setosa.petal_length,
    y=setosa.petal_width,
    fill=True
)

plt.title("KDE Plot of Sentosa Petal Dimensions")
plt.xlabel("Petal Length")
plt.ylabel("Petal Width")
plt.show()
```
